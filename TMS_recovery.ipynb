```python
N = 30
delta = 0.3
gamma = 0.1
gkp_qunaught = gkp_qunaught_state(N, delta)
loss = boson_loss_channel(0.1, N)
encoding = gkp_encoding(N, delta)
lossy_encoding = choi_compound(encoding, loss)
```
